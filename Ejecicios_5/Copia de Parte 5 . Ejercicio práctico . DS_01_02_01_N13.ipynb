## Ejercicio 2
Haz un programa que detecte todos los duplicados de un elemento en una lista:
* Utiliza una variable duplicado para asignarle el valor del elemento del que queremos encontrar todos sus duplicados.
* El programa debe devolver todos los índices de los elementos duplicados.
* Aplícalo para encontrar los duplicados de "un", "es" y "binario" de:  

>```ejer_3 = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]```


`````python
lista = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]

# elementos de los que encontra indice
duplicados = ["un", "es", "binario"]
for duplicado in duplicados:
    indices = []
    for i in range(len(lista)):
        # Convertimos ambos a minúsculas para que coincidan sin importar mayúsculas
        if lista[i].lower() == duplicado.lower():
            indices.append(i)
    print(f"Los índices para '{duplicado}' son: {indices}")
`````

## Ejercicio 10
Convierte el programa del ejercicio 2 en una función al que se le pase el valor a buscar como un argumento posicional

`````python
lista = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]

def buscar_duplicados(lista_busqueda, elemento):
    indices = []
    for i in range(len(lista_busqueda)):
        if lista_busqueda[i].lower() == elemento.lower():
            indices.append(i)
    return indices

# Demostración del funcionamiento
for palabra in ["un", "es", "binario"]:
    resultado = buscar_duplicados(lista, palabra)
    print(f"Los índices para '{palabra}' son: {resultado}")
`````

## Ejercicio 13
Escribe una función que compruebe si un DNI es válido, devolviendo `True` en caso afirmativo.  
En primer lugar, el DNI debe tener entre 7 y 8 dígitos numéricos.  
Para saber si la letra del DNI es correcta puedes utilizar el siguiente fragmento de código:
>```Python
>DNI= 55555555
>palabra='TRWAGMYFPDXBNJZSQVHLCKE'
>print('La letra del DNI es: ', palabra[DNI%23])
>````


`````python
def validar_dni(dni_completo):
    # Eliminamos posibles espacios en blanco
    dni_completo = dni_completo.strip().upper()

    # Separamos parte numerica de la letra
    parte_numerica = dni_completo[:-1]
    letra_introducida = dni_completo[-1]

    # comprobamos que la parte numerica tenga entre 7 y 8 dígitos y sea numerica
    if not (7 <= len(parte_numerica) <= 8) or not parte_numerica.isdigit():
        return False

    # Convertimos a numero para aplicar la formula de la letra
    dni_numero = int(parte_numerica)
    palabra = 'TRWAGMYFPDXBNJZSQVHLCKE'
    letra_correcta = palabra[dni_numero % 23]

    # El DNI es valido si la letra introducida coincide con la calculada
    return letra_introducida == letra_correcta
# Pruebas de funcionamiento
ejemplos_dni = ["55555555J", "12345678A", "1234567Z"]
for dni in ejemplos_dni:
    resultado = validar_dni(dni)
    print(f"¿El DNI {dni} es válido?: {resultado}")
`````
